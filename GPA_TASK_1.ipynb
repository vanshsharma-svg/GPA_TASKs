# Load DataSet `play_store_data`

```python
import pandas as pd

play_store_data = pd.read_csv(
    r"C:\Users\Vansh Sharma\Downloads\Play Store Data (1).csv"
)

pd.DataFrame(play_store_data.head())
```

## Data Cleaning from 

- Removed corrupted rows from the dataset.
- Converted date columns into datetime format.
- Converted Reviews into numeric format.
- Handled missing values using forward fill and mode imputation.
- Prepared data for further analysis.

```python
 # 1. Remove the corrupted row
play_store_data = play_store_data[play_store_data["Installs"] != "Free"]

# 2. Remove commas and plus signs
# play_store_data["Installs"] = (
#     play_store_data["Installs"]
#     .str.replace(",", "", regex=False)
#     .str.replace("+", "", regex=False)
# )

# # 3. Convert to integer
# play_store_data["Installs"] = play_store_data["Installs"].astype(int)

# 4. Convert Last Updated to datetime
play_store_data["Last Updated"] = pd.to_datetime(
    play_store_data["Last Updated"],
    errors="coerce"
)

# covert into int
play_store_data["Reviews"] = play_store_data["Reviews"].astype(int)


# fwd filling (Rating)
play_store_data["Rating"] = play_store_data["Rating"].ffill()
play_store_data["Current Ver"] = play_store_data["Current Ver"].ffill()
play_store_data["Android Ver"] = play_store_data["Android Ver"].ffill()
play_store_data["Type"] = play_store_data["Type"].fillna(
    play_store_data["Type"].mode()[0]
)


```

# Fill `np.NAN` with median

```python
play_store_data["Size"] = play_store_data["Size"].fillna(
    play_store_data["Size"].median()
)
```

# Remove Duplicates

```python
play_store_data = play_store_data.drop_duplicates()
```

## Data Filtering & Aggregation

- Filtered apps with size ≥ 10 MB and January updates.
- Grouped data by Category.
- Calculated average rating, total reviews, and total installs.
- Removed categories with average rating below 4.0.
- Selected top 10 categories based on total installs.

```python
filtered = play_store_data[
    (play_store_data["Size"] >= 10) &
    (play_store_data["Last Updated"].dt.month == 1)
]


filtered_summary = (
    filtered.groupby("Category", as_index=False)
    .agg({
        "Rating":"mean",
        "Reviews":"sum",
        "Installs":"sum"
    })
)


# Remove categories having average rating below 4
filtered_summary = filtered_summary[
    filtered_summary["Rating"] >= 4
]


# Top 10 categories by installs
filtered_summary = (
    filtered_summary
    .sort_values("Installs", ascending=False)
    .head(10)
)


filtered_summary["Rating"] = filtered_summary["Rating"].round(2)

filtered_summary
```

## Visualization

- Created a grouped bar chart to compare Average Rating and Total Reviews.
- Used dual y-axis due to different scales of metrics.
- Added time-based visibility condition:
  - Graph displays only between 3 PM IST and 5 PM IST.
  - Hidden outside the specified time range.

```python
# Current IST time
ist = pytz.timezone("Asia/Kolkata")
current_time = datetime.now(ist).time()


# Allowed time range
start_time = datetime.strptime("15:00", "%H:%M").time()
end_time = datetime.strptime("17:00", "%H:%M").time()


# Show graph only between 3 PM - 5 PM IST
if start_time <= current_time <= end_time:

    x = np.arange(len(filtered_summary["Category"]))
    width = 0.35

    fig, ax1 = plt.subplots(figsize=(14,6))

    # Average Rating
    ax1.bar(
        x - width/2,
        filtered_summary["Rating"],
        width,
        label="Average Rating"
    )

    ax1.set_xlabel("Category")
    ax1.set_ylabel("Average Rating")
    ax1.set_xticks(x)
    ax1.set_xticklabels(
        filtered_summary["Category"],
        rotation=90
    )


    # Total Reviews
    ax2 = ax1.twinx()

    ax2.bar(
        x + width/2,
        filtered_summary["Reviews"],
        width,
        label="Total Reviews"
    )

    ax2.set_ylabel("Total Reviews")


    plt.title(
        "Top 10 App Categories: Average Rating vs Total Reviews"
    )

    fig.legend(
        loc="upper right",
        bbox_to_anchor=(0.9,0.9)
    )

    plt.grid(axis="y")
    plt.show()


else:
    print("Graph is available only between 3 PM IST and 5 PM IST")
```

# `KPIs`

```python
# KPIs

total_categories = filtered_summary["Category"].nunique()

highest_install_category = filtered_summary.loc[
    filtered_summary["Installs"].idxmax(),
    "Category"
]

avg_rating = filtered_summary["Rating"].mean().round(2)

total_reviews = filtered_summary["Reviews"].sum()

total_installs = filtered_summary["Installs"].sum()

print("Total Categories:", total_categories)
print("Highest Install Category:", highest_install_category)
print("Average Rating:", avg_rating)
print("Total Reviews:", total_reviews)
total_installs = filtered_summary["Installs"].count()
print("Total Install:",total_installs)
```
